# B100 Intelligence - EDA Exploration

This notebook is used for Exploratory Data Analysis (EDA) to visualize trends and prototype scoring metrics before they are formalized into the `scoring_engine.py` pipeline.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_theme(style="whitegrid")
```

### 1. Load Data
Loading cleaned data directly from the CSV warehouse backups or via PostgreSQL.

```python
# Load clean CSVs for exploration
pl = pd.read_csv('../data/clean/profitandloss.csv')
bs = pd.read_csv('../data/clean/balancesheet.csv')

print(f"Loaded {len(pl)} P&L records and {len(bs)} Balance Sheet records.")
```

### 2. Profitability Trends
Visualizing the net profit growth trajectory for top companies over the last 10 years.

```python
# Example: Plot Net Profit for a sample of companies
sample_companies = ['TCS', 'HDFCBANK', 'INFY']
df_sample = pl[pl['company_id'].isin(sample_companies)]

plt.figure(figsize=(10, 6))
sns.lineplot(data=df_sample, x='fiscal_year', y='net_profit', hue='company_id', marker='o')
plt.title('Net Profit Trends (Sample Nifty 100)')
plt.ylabel('Net Profit (Cr)')
plt.xlabel('Fiscal Year')
plt.show()
```
